```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('heart.csv')
```

```python
df.head()
```

```python
df.corr(numeric_only=True)
```

## Distribution Plot for Numeric Columns

```python
sns.displot(df['Age'],kde=True)
```

```python
sns.distplot(df['Age'],hist=True,kde=False)
```

```python
sns.distplot(df['Age'],bins=20)
```

```python
sns.distplot(df['RestingBP'],bins=30,color='red')
```

```python
sns.distplot(df['MaxHR'],bins=30,color='red')
```

## Pie Charts for Categorical Ones

```python
df.groupby('Sex').size().plot(kind='pie',autopct='%.1f%%')
```

```python
df.groupby('ChestPainType').size().plot(kind='pie',autopct='%.1f%%')
```

```python
df.groupby('RestingECG').size().plot(kind='pie',autopct='%.1f%%')
```

```python
df.groupby('ST_Slope').size().plot(kind='pie',autopct='%.1f%%')
```

```python
df.groupby('HeartDisease').size().plot(kind='pie',autopct='%.1f%%')
```

### Violin Plot

```python
sns.violinplot(data=df,y='Sex',x='HeartDisease')
```

```python
sns.violinplot(data=df,y='Age',x='HeartDisease')
```

```python
sns.violinplot(data=df,y='Cholesterol',x='HeartDisease')
```

## HeatMap or Correlation

```python
sns.heatmap(df.corr(numeric_only=True))
```

## Correlation - (JointPlot)

```python
sns.jointplot(x='Age',y='MaxHR',data=df,kind='hex')
```

```python
sns.jointplot(x='Age',y='MaxHR',data=df,kind='reg')
```

```python
sns.jointplot(x='HeartDisease',y='MaxHR',data=df,kind='reg')
```

```python
sns.jointplot(x='Cholesterol',y='Oldpeak',data=df,kind='reg') #No Correlation
```
