```python
res = session.get('https://en.wikipedia.org/wiki/Mahatma_Gandhi')
soup=BeautifulSoup(res.text,'html.parser')
heading=soup.find('h1').text
corpus =''
for i in soup.find_all('p'):
    corpus+=i.text
    corpus+='\n'
corpus=corpus.strip()
```

```python
for i in range(24,176):
    corpus = corpus.replace('['+str(i)+']','')
    
```

```python
print(corpus)
```

```python
fd = open(heading+'.txt','w',encoding='utf-8')
fd.write(corpus)
fd.close()
```
